```python
import os
import numpy as np
import pandas as pd
CHUNKSIZE = 500
```

```python
data = pd.DataFrame()
data['id'] = np.arange(10211, 21055)
data['id'] = data['id'].apply(lambda x: f'{x:07d}')
data['location'] = 'B175'
data['status'] = 'Stored'
data['owner'] = 'CLJ'
data
```

```python
output_dir = '/root/cernbox/equipment_lists/active_hv_dividers'
filepath = os.path.join(
    output_dir,
    'generated_hv_dividers_id_{i}.csv'
)

for i, chunk_start in enumerate(range(0, len(data), CHUNKSIZE)):
    data_to_save = data.iloc[chunk_start:chunk_start+CHUNKSIZE]
    chunkfilepath = filepath.format(i=f'{i:03d}')
    data_to_save.to_csv(chunkfilepath, index=False)
```
