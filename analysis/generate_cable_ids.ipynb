```python
import os
import numpy as np
import pandas as pd
from utils import dump_in_batches
```

# LV Cables

```python
def make_lv_cable_id(module, start, finish, preproduction, harness):
    harness_number = {
    'girder': 1,
    'finger': 2
    }
    module_number = {
        'EB': 27,
        'LB': 26
    }
    ids = [
        f'{module_number[module]}{cable_num:03d}{int(preproduction)}{harness_number[harness]}'
        for cable_num in range(start, finish)
    ]
    return ids
```

```python
lb_lv_cables_df = pd.DataFrame()
lb_lv_cables_df['id'] = \
    make_lv_cable_id('LB', 1, 5, True, 'girder') + \
    make_lv_cable_id('LB', 1, 5, True, 'finger') + \
    make_lv_cable_id('LB', 5, 141, False, 'girder') + \
    make_lv_cable_id('LB', 5, 141, False, 'finger')
lb_lv_cables_df['location'] = 'B175'
lb_lv_cables_df['status'] = 'Stored'
lb_lv_cables_df['owner'] = 'CLJ'
dump_in_batches(
    df=lb_lv_cables_df,
    filename='generated_lb_lv_cables',
    output_dir='/root/cernbox/equipment_lists/lv_cables',
    chunksize=200
)
```

```python
eb_lv_cables_df = pd.DataFrame()
eb_lv_cables_df['id'] = \
    make_lv_cable_id('EB', 1, 5, True, 'girder') + \
    make_lv_cable_id('EB', 1, 5, True, 'finger') + \
    make_lv_cable_id('EB', 5, 141, False, 'girder') + \
    make_lv_cable_id('EB', 5, 141, False, 'finger')
eb_lv_cables_df['location'] = 'B175'
eb_lv_cables_df['status'] = 'Stored'
eb_lv_cables_df['owner'] = 'CLJ'
dump_in_batches(
    df=eb_lv_cables_df,
    filename='generated_eb_lv_cables',
    output_dir='/root/cernbox/equipment_lists/lv_cables',
    chunksize=200
)
```
